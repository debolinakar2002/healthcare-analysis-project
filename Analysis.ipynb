```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('data/healthcare_dataset.csv')
df.head()
```

```python
df.info()
df.describe()
```

```python
df['Date of Admission']=pd.to_datetime(df['Date of Admission'])
df['Discharge Date']=pd.to_datetime(df['Discharge Date'])

df['Stay_Duration']=(df['Discharge Date'] - df['Date of Admission']).dt.days
```

```python
df['Medical condition'].value_counts()
```

```python
df['Medical Condition'].value_counts()
```

```python
sns.countplot(x='Medical Condition', data=df)
plt.xticks(rotation=45)
plt.show()
```

```python
sns.histplot(df['Age'],bins=20)
plt.title('Age Distribution')
plt.show()
```

```python
df['Gender'].value_counts().plot(kind='pie',autopct='%1.1f%%')
plt.show()
```

```python
df['Stay_Duration'].mean()
```

Insights
- Majority of patients fall in middle age group (40-70 years)
- Certain medical conditions are more frequent like Arthritis, Diabetes, and Hypertension
- Billing amount varies widely among patients depending on treatment
- Average hospital stay is around 15.5 days
- Gender distribution is relatively balanced (50% male, 50% female)

Conclusion
- This project analyzes healthcare data to understand patient demographics, medical condition and hospital trends. It helps in understanding common diseases, understanding patients age distribution and estimating hospital stay duration , which can support better healthcare decision making.
